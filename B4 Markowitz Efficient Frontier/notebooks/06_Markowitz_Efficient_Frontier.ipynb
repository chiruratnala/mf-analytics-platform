## Interpretation
- **Minimum Volatility** identifies the lowest-risk portfolio under the long-only constraint.
- **Maximum Sharpe** identifies the portfolio with the strongest return per unit of annualised volatility.
- The efficient frontier represents portfolios that are not dominated by another feasible portfolio with lower risk and higher return.
- Results are historical estimates, not guaranteed future performance.
